```python
#Que9- After removing duplicates and converting status to lower case, return status and loans (the number of loans) for each status.

gold_loans = gold_loans.drop_duplicates()
gold_loans['status'].unique()
gold_loans['status'] = gold_loans['status'].str.strip().str.lower()
# answer = gold_loans.groupby('status').size().reset_index(name='loans')
# answer = gold_loans.groupby('status')['loan_id'].nunique().reset_index(name='loans')
answer = gold_loans['status'].value_counts().reset_index(name='loans')

answer
```

```python
#Que10- How many different raw values does the purity column contain?
gold_loans.head()

gold_loans['purity'].unique()
```

```python
#Que11- After removing duplicates and converting purity to a karat number (916 = 22, 833 = 20, 750 = 18, 999 = 24), return karat and loans for each karat.

gold_loans['purity'] = gold_loans['purity'].str.replace('K','').str.replace('T','').str.strip().astype(int)

gold_loans['karat']  = gold_loans['purity'].replace({916 : 22, 833 : 20, 750 : 18, 999 : 24})

answer  = gold_loans.groupby('karat').size().reset_index(name='loans')

answer
```

```python
#Que31- What is the total outstanding principal (₹) of all assessable active loans? Outstanding = principal − principal repayments.

gold_loans.head(5)
repayments.head(5)

paid = repayments[(repayments['payment_type']=='principal')&repayments['loan_id'].isin(active_loans['loan_id'])]

outstanding = active_loans['principal'].sum()-paid['amount'].sum()

print(outstanding)
```

```python
#Que32- What is the average current LTV (%) of assessable active loans? Round to 2 decimals.

gold_loans.head(5)
rates.head()
active_loans.head()

rate_asof = rates.loc[rates['rate_date'] == '2026-10-15', 'rate_24k_per_gram'].iloc[0]

paid_per_loan = paid.groupby('loan_id')['amount'].sum()

active_loans['outstanding_loan']=active_loans['principal']-active_loans['loan_id'].map(paid_per_loan).fillna(0)
active_loans['collateral'] = active_loans['gold_weight_gm'] * active_loans['karat'] / 24 * rate_asof
active_loans['current_ltv'] = active_loans['outstanding_loan'] / active_loans['collateral'] * 100

avg_current_ltv = active_loans['current_ltv'].mean().round(2)

print('Average Current LTV :' , avg_current_ltv)
```

```python
#Que33- How many assessable active loans are above the 75% LTV cap?
active_loans['current_ltv']
loans_above_75_ltv = (active_loans['current_ltv'] > 75).sum()
print('Total active Loans above 75% LTV : ',loans_above_75_ltv)
```

```python
#Que34-What is the outstanding principal (₹) of loans above 75% LTV?
principal_above_75_ltv = active_loans[active_loans['current_ltv'] > 75]['principal'].sum()
print('Total principal above 75% LTV : ',principal_above_75_ltv)
```

```python
#Que35-What percentage of assessable active loans are in breach? Round to 2 decimals.
active_loans.head()
active_loans['breach_loans'] = active_loans['current_ltv'] > 75
breach_loans_pct = round(len(active_loans[active_loans['current_ltv'] > 75])/len(active_loans)*100,2)
print('Total active loan percentage in breach is :',breach_loans_pct,'%')
```

```python
#Que36-Return scheme, breach_loans and active_loans (assessable active loans) for each scheme.
active_loans.head()
answer = active_loans.groupby('scheme').agg(breach_loans = ('breach_loans','sum'),active_loans = ('loan_id','count')).reset_index()
answer
```

```python
#Que37-Return state and breach_outstanding (₹, rounded to 0) for every state with breaches, highest first.
active_loans.head()
branches.head()

merged = active_loans[active_loans['breach_loans']].merge(branches,on='branch_id',how='left')

answer = merged.groupby('state')['outstanding_loan'].sum().round(0).sort_values(ascending=False).reset_index(name='breach_outstanding')

answer
```

```python
#Que38-For breaching loans, return month (text 'YYYY-MM' of the disbursal date) and breach_loans, in month order.
breach_loans_df  = active_loans[active_loans['current_ltv']>75]

breach_loans_df['month']  = breach_loans_df['disbursal_date'].dt.strftime('%Y-%m')

answer = breach_loans_df.groupby('month')['loan_id'].count().reset_index(name='breach_loans')

answer
```

```python
#Que39-Return the top 10 branches by breached outstanding: branch_id, branch_name, breach_outstanding (₹, rounded to 0), highest first.
merged.head()

answer = merged.groupby(['branch_id','branch_name'])['outstanding_loan'].sum().round(0).sort_values(ascending=False).head(10).reset_index(name='breach_outstanding')

answer
```

```python
#Que40-What percentage of assessable active Express loans are in breach? Round to 1 decimal.
breach_loans_df.head()

total_active_express_loans = active_loans[active_loans['scheme']=='Express']
total_active_breach_express_loans = breach_loans_df[breach_loans_df['scheme']=='Express']

answer = round(len(total_active_breach_express_loans)/len(total_active_express_loans)*100,1)

answer
```

```python
#Que41-Return region, active_loans and outstanding (₹, rounded to 0) for assessable active loans in each region.

active_loans.head()
branches.head()

all_merged = active_loans.merge(branches,on='branch_id',how='left')

answer = all_merged.groupby('region').agg(active_loans = ('loan_id','count'),outstanding = ('outstanding_loan','sum')).reset_index()

answer['outstanding'] = answer['outstanding'].round(0)
answer
```

```python
#Que42-For loans disbursed from 1 July to 30 September 2026 (after removing duplicates), return scheme and avg_ltv_at_disbursal (rounded to 2).

gold_loans.head()
gold_loans.duplicated().sum()

df = gold_loans[(gold_loans['disbursal_date']>='2026-07-01') & (gold_loans['disbursal_date']<='2026-09-30')]

answer = df.groupby('scheme')['ltv_at_disbursal_pct'].mean().round(2).reset_index(name='avg_ltv_at_disbursal')
answer
```

```python
#Que43-How many active loans (after removing duplicates) are past their maturity date (disbursal date + tenure months) on 15 October 2026?

active_loans.head()

active_loans['maturity_date'] = active_loans.apply(
    lambda r: r['disbursal_date'] + pd.DateOffset(months=int(r['tenure_months'])),
    axis=1
)

answer = (active_loans['maturity_date'] < '2026-10-15').sum()
print(answer)
```

```python
#Que46-If the gold rate falls a further 5% from the assessment-date rate, how many assessable active loans would be above 75% LTV?

active_loans['collateral_after_fall'] = active_loans['collateral'] * 0.95

active_loans['ltv_after_fall'] = active_loans['outstanding_loan'] / active_loans['collateral_after_fall'] * 100

answer = (active_loans['ltv_after_fall'] > 75).sum()
print(answer)
```

```python
#Que47-How much principal (₹, rounded to 0) must be repaid in total to bring every breaching loan back to exactly 75% LTV?
breach_loans_df.head()

breach_loans_df['repay_needed'] = breach_loans_df['outstanding_loan'] - 0.75 * breach_loans_df['collateral']

answer = round(breach_loans_df['repay_needed'].sum(), 0)
print(answer)
```

```python
#Que49-For each region, return the branch with the highest breach rate (% of its assessable active loans in breach): region, branch_name, breach_rate_pct (rounded to 2).
all_merged.head()

rate = (all_merged.groupby(['region', 'branch_id', 'branch_name'])['breach_loans'].mean() * 100
).reset_index(name='breach_rate_pct')

rate['breach_rate_pct'] = rate['breach_rate_pct'].round(2)

answer = rate.sort_values('breach_rate_pct', ascending=False).drop_duplicates('region')

answer = answer[['region', 'branch_name', 'breach_rate_pct']]

answer
```

```python
#Que50-For breaching loans disbursed in 2026, return month ('YYYY-MM'), breach_outstanding (rounded to 0) and cumulative_outstanding (running total, rounded to 0), in month order.

breach_loans_df = breach_loans_df[breach_loans_df['disbursal_date'].dt.year == 2026]
breach_loans_df['month'] = breach_loans_df['disbursal_date'].dt.strftime('%Y-%m')

answer = breach_loans_df.groupby('month')['outstanding_loan'].sum().round(0).reset_index(name='breach_outstanding')
answer['cumulative_outstanding'] = answer['breach_outstanding'].cumsum()

answer
```
